```python
claims_data = [
    ("CLM-001", "MRN-100234", 1200.50, "Approved"),
    ("CLM-002", "MRN-100235", 8750.25, "Pending"),
    ("CLM-003", "MRN-100236", 340.00, "Denied"),
    ("CLM-004", "MRN-100237", 4300.75, "Approved"),
    ("CLM-005", "MRN-100238", 6100.00, "Pending"),
]
print("Approved Claims:")
for claim_id, mrn, amount, status in claims_data:
    if status == "Approved":
        print(f"Claim {claim_id} | Patient {mrn} | Amount ${amount}")

# Q2 - your list comprehension here
pending_amounts = [amount for claim_id, mrn, amount, status in claims_data if status == "Pending"]
print("Pending amounts:", pending_amounts)
print("Pending total: $", sum(pending_amounts))
```
