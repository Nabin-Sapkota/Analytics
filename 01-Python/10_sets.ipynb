```python
# Raw diagnosis codes from a claims file — full of duplicates
raw_codes = ["I10", "E11.9", "I10", "J18.9", "I10", "E11.9", "I21.9", "J18.9"]

# convert to set - duplicates gone instantly
unique_codes = set(raw_codes)
print(unique_codes)
print(len(unique_codes))
```

```python
# Patients seen in January
january_patients = {"MRN-001", "MRN-002", "MRN-003", "MRN-004"}

# Patients seen in February
february_patients = {"MRN-003", "MRN-004", "MRN-005", "MRN-006"}

either_month = january_patients | february_patients
print(either_month)
# {'MRN-001', 'MRN-002', 'MRN-003', 'MRN-004', 'MRN-005', 'MRN-006'}

both_months = january_patients & february_patients
print(both_months)
# {'MRN-003', 'MRN-004'}
```

```python
approved_facilities = {"FAC-001", "FAC-002", "FAC-003"}

facility = "FAC-007"
if facility not in approved_facilities:
    print(f"{facility} is not an approved network facility")
```

```python
# Claims submitted this month from two regions
region_a_patients = {"MRN-001", "MRN-002", "MRN-003", "MRN-004", "MRN-005"}
region_b_patients = {"MRN-004", "MRN-005", "MRN-006", "MRN-007"}

# Diagnosis codes in a messy claims extract
raw_diagnoses = ["I10", "E11.9", "I10", "J18.9", "E11.9", 
                 "I21.9", "I10", "J18.9", "E10.9", "I21.9"]

both_regions = region_a_patients & region_b_patients
print(both_regions)
a_not_b = region_a_patients - region_b_patients
print(a_not_b)
unique_raw_diagnoses_code = set(raw_diagnoses)
print(unique_raw_diagnoses_code)
code_to_find = ["I10", "Z99.9"]
for code in code_to_find:
    if code in unique_raw_diagnoses_code:
        print(f"{code} : Code Found")
    else:
        print(f"{code} : Code not found")
```
