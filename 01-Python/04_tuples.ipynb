```python
# A single diagnosis record — code + description + category
# This should never be modified mid-script
diagnosis = ("I10", "Essential Hypertension", "Cardiovascular")

# Patient identity record — MRN + DOB
patient_record = ("MRN-100234", "1965-03-12", "John Smith")

# Allowed visit types — fixed reference values
allowed_visit_types = ("Inpatient", "Outpatient", "ER", "Urgent Care")

# Accessing tuple values 
print(diagnosis[0])
print(diagnosis[1])
print(diagnosis[2])
```

```python
# Unpacking into meaningful variable names
icd_code, description, category = diagnosis 
print(icd_code)
print(description)
print(category)
```

```python
# this is what sql query looks like in python, each row comes back as a tuple 
claims_data = [
    ("CLM-001", "MRN-100234", 1200.50, "Approved"),
    ("CLM-002", "MRN-100235", 8750.25, "Pending"),
    ("CLM-003", "MRN-100236", 340.00, "Denied"),
    ("CLM-004", "MRN-100237", 4300.75, "Approved")
]

# loop and unpack each row 
for claim_id, mrn, amount, status in claims_data : 
    print(f"Claim {claim_id} for patient {mrn} : ${amount} - {status}")
```

```python
# This is not a tuple - it's just a string in parentheses 
not_a_tuple = ("I10")
print(type(not_a_tuple)) # <class 'str'>

#This is a tuple -notice the trailing comm
real_tuple = ("I10",)
print(type(real_tuple)) # <class 'tuple'>
```
