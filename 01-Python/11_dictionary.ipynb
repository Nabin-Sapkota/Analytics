```python
# A patient record — key is the field name, value is the data
patient = {
    "mrn":          "MRN-100234",
    "name":         "John Smith",
    "dob":          "1965-03-12",
    "gender":       "M",
    "payer":        "Medicare",
    "plan_type":    "FFS"
}

# Access by key
print(patient["name"])      # John Smith
print(patient["payer"])     # Medicare
```

```python
# add a new field 
patient["zip_code"] = "10001"

# Update an existing field 
patient["payer"] = "Medicaid"

# Delete a field 
del patient["zip_code"]

# safe access - use .get() to void KeyError on missing keys 
# this is critical when processing messy claims files 

print(patient.get("phone", "No phone on file")) # no phone on file

# check if key exists 
if "dob" in patient: 
    print("Date of birth recorded") 

# Loop through all fields 

for key, value in patient.items(): 
    print(f" {key} : {value}")

```

```python
icd10_lookup = {
    "I10":    "Essential Hypertension",
    "E11.9":  "Type 2 Diabetes without complications",
    "J18.9":  "Pneumonia, unspecified",
    "I21.9":  "Acute Myocardial Infarction, unspecified",
    "S06.0":  "Concussion",
    "E10.9":  "Type 1 Diabetes without complications"
}

# Look up a single code
code = "I10"
print(icd10_lookup[code])  # Essential Hypertension

# Look up codes from a claims list safely
claim_codes = ["I10", "E11.9", "Z99.9"]  # Z99.9 not in lookup

for code in claim_codes:
    description = icd10_lookup.get(code, "Unknown code — needs review")
    print(f"{code}: {description}")
```

```python
# Multiple patients — outer key is MRN, value is patient record
patients_db = {
    "MRN-001": {"name": "John Smith",   "age": 58, "payer": "Medicare"},
    "MRN-002": {"name": "Maria Garcia", "age": 41, "payer": "BlueCross"},
    "MRN-003": {"name": "James Lee",    "age": 65, "payer": "Medicaid"},
}

# Look up a specific patient
print(patients_db["MRN-002"]["name"])   # Maria Garcia
print(patients_db["MRN-002"]["payer"])  # BlueCross

# Loop all patients
for mrn, info in patients_db.items():
    print(f"{mrn} — {info['name']} — {info['payer']}")
```
